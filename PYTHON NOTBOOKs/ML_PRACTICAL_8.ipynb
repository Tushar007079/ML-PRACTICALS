####**Drug Prediction Data**

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

data = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/ML_7/p6_drug200.csv")
print("\nData Preview : \n")
print(data.head(10))
```

```python
print("\nDataset Information:\n")
data.info()

label_encoder = LabelEncoder()
for col in ['Sex', 'BP', 'Cholesterol', 'Drug']:
    data[col] = label_encoder.fit_transform(data[col])

# Define features (X) and target (y)
X = data.drop('Drug', axis=1)
y = data['Drug']

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("\nData preprocessing completed.")
print(f"Training set size: {X_train.shape}, Testing set size: {X_test.shape}")
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling applied for improved model convergence.")
```

```python
from imblearn.over_sampling import SMOTE

# Use SMOTE to oversample the minority classes
smote = SMOTE(random_state=42)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train_scaled, y_train)

print("Applied SMOTE to balance the dataset.")
```

```python
models = {
    '📊 Logistic Regression': LogisticRegression(solver='saga', max_iter=3000, random_state=42),
    '🧑‍🤝‍🧑 K-Nearest Neighbors': KNeighborsClassifier(),
    '🌳 Decision Tree': DecisionTreeClassifier(),
    '🧠 Artificial Neural Network': MLPClassifier(hidden_layer_sizes=(10,), max_iter=2000, random_state=42, early_stopping=True, learning_rate_init=0.001)
}

metrics = {}

# Train and evaluate models
for model_name, model in models.items():
    model.fit(X_train_balanced, y_train_balanced)  # Use balanced dataset
    y_pred = model.predict(X_test_scaled)

    # Metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted', zero_division=1)  # Handle undefined metrics
    recall = recall_score(y_test, y_pred, average='weighted')
    f1 = f1_score(y_test, y_pred, average='weighted')
    cm = confusion_matrix(y_test, y_pred)

    # Store metrics in dict
    metrics[model_name] = {
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1_score': f1,
        'confusion_matrix': cm
    }

print("Training completed!")
```

```python
from tabulate import tabulate

#Results
print("\nModel Performance Metrics:\n")
for model_name, metric in metrics.items():
    print(f"\n{model_name} : \n")
    print(f"Confusion Matrix : \n{metric['confusion_matrix']}\n")
    print("-" * 80)

comparison_table = [
    [model,
     f"{metrics[model]['accuracy']:.4f}",
     f"{metrics[model]['precision']:.4f}",
     f"{metrics[model]['recall']:.4f}",
     f"{metrics[model]['f1_score']:.4f}"]
    for model in metrics
]

print("\nComparison of Models : \n")
print(tabulate(
    comparison_table,
    headers=["Model", "Accuracy", "Precision", "Recall", "F1 Score"],
    tablefmt="fancy_grid"
))
```

####**Credit Card Fraud Detection Data**

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
df_fraud = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/ML_7/p5_creditcard.csv')
print("\nData Preview : \n", df_fraud.head(10))
print("\nDataset Info : \n")
df_fraud.info()
```

```python
# Preprocessing
X_fraud = df_fraud.drop('default.payment.next.month', axis=1)
y_fraud = df_fraud['default.payment.next.month']

# Scaling
scaler = StandardScaler()
X_fraud_scaled = scaler.fit_transform(X_fraud)

# Splitting
X_train_fraud, X_test_fraud, y_train_fraud, y_test_fraud = train_test_split(
    X_fraud_scaled, y_fraud, test_size=0.3, random_state=42
)
print(f"Training Data Shape: {X_train_fraud.shape}, Test Data Shape: {X_test_fraud.shape}")
```

```python
# Train and Evaluate Models
models_fraud = {
    '📊 Logistic Regression': LogisticRegression(max_iter=2000),
    '🧑‍🤝‍🧑 K-Nearest Neighbors': KNeighborsClassifier(),
    '🌳 Decision Tree': DecisionTreeClassifier(),
}

metrics_fraud = {}

for model_name, model in models_fraud.items():
    model.fit(X_train_fraud, y_train_fraud)
    y_pred_fraud = model.predict(X_test_fraud)

    accuracy = accuracy_score(y_test_fraud, y_pred_fraud)
    precision = precision_score(y_test_fraud, y_pred_fraud, zero_division=0)
    recall = recall_score(y_test_fraud, y_pred_fraud, zero_division=0)
    f1 = f1_score(y_test_fraud, y_pred_fraud, zero_division=0)
    cm = confusion_matrix(y_test_fraud, y_pred_fraud)

    metrics_fraud[model_name] = {
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1_score': f1,
        'confusion_matrix': cm
    }

print("\nPerformance Metrics for Logistic Regression, KNN, and Decision Tree:")
for model_name, metric in metrics_fraud.items():
    print(f"\nModel : {model_name}\n")
    print(f"  Accuracy: {metric['accuracy']:.4f}")
    print(f"  Precision: {metric['precision']:.4f}")
    print(f"  Recall: {metric['recall']:.4f}")
    print(f"  F1 Score: {metric['f1_score']:.4f}")
    print(f"  \nConfusion Matrix:\n{metric['confusion_matrix']}\n")
    print("-" * 80)
```

```python
# ANN Model
ann_model = Sequential([
    Dense(64, input_dim=X_train_fraud.shape[1], activation='relu'),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])

ann_model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

ann_model.fit(X_train_fraud, y_train_fraud, epochs=50, batch_size=32, verbose=1)

# ANN Predictions
y_pred_fraud_ann = (ann_model.predict(X_test_fraud) > 0.5).astype("int32")

accuracy_ann = accuracy_score(y_test_fraud, y_pred_fraud_ann)
precision_ann = precision_score(y_test_fraud, y_pred_fraud_ann, zero_division=0)
recall_ann = recall_score(y_test_fraud, y_pred_fraud_ann, zero_division=0)
f1_ann = f1_score(y_test_fraud, y_pred_fraud_ann, zero_division=0)
cm_ann = confusion_matrix(y_test_fraud, y_pred_fraud_ann)

metrics_fraud['🤖 Artificial Neural Network (Keras)'] = {
    'accuracy': accuracy_ann,
    'precision': precision_ann,
    'recall': recall_ann,
    'f1_score': f1_ann,
    'confusion_matrix': cm_ann
}
```

```python
from tabulate import tabulate

# Summary of all models
summary_data = []

for model_name, metric in metrics_fraud.items():
    summary_data.append([
        model_name,
        metric['accuracy'],
        metric['precision'],
        metric['recall'],
        metric['f1_score']
    ])

headers = ["Model", "Accuracy", "Precision", "Recall", "F1 Score"]

print("\nFinal Performance Metrics (Tabular Form):")
print(tabulate(summary_data, headers=headers, tablefmt="fancy_grid", floatfmt=".4f"))
```

####**Metro Interstate Traffic Volume Data**

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tabulate import tabulate
```

```python
data = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/ML_7/p3_metro.csv')
print("\nData Preview : \n\n", data.head(10))
print()
print("-" * 100)
print("\nDataset Info : \n")
data.info()
print()

data.drop(['holiday', 'weather_main', 'weather_description', 'date_time'], axis=1, inplace=True)

data.fillna(data.mean(), inplace=True)

print("-" * 100)

print("\nProcessed Data : \n\n", data.head(10))

X = data.drop('traffic_volume', axis=1)
y = data['traffic_volume']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Multi-linear regression

linear_model = LinearRegression()

linear_model.fit(X_train, y_train)

y_pred_linear = linear_model.predict(X_test)

mse_linear = mean_squared_error(y_test, y_pred_linear)
r2_linear = r2_score(y_test, y_pred_linear)
```

```python
# Transform features to polynomial degree 2
poly = PolynomialFeatures(degree=2)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)

# Polynomial Regression
poly_model = LinearRegression()
poly_model.fit(X_train_poly, y_train)

y_pred_poly = poly_model.predict(X_test_poly)

mse_poly = mean_squared_error(y_test, y_pred_poly)
r2_poly = r2_score(y_test, y_pred_poly)
```

```python
# ANN

ann_model = Sequential()
ann_model.add(Dense(64, input_dim=X_train.shape[1], activation='relu'))
ann_model.add(Dense(32, activation='relu'))
ann_model.add(Dense(1, activation='linear'))

ann_model.compile(optimizer='adam', loss='mse', metrics=['mse'])

ann_model.fit(X_train, y_train, epochs=50, batch_size=32, verbose=1)

y_pred_ann = ann_model.predict(X_test).flatten()

mse_ann = mean_squared_error(y_test, y_pred_ann)
r2_ann = r2_score(y_test, y_pred_ann)
```

```python
# Results

results = [
    ['🔥 Multi-Linear Regression', mse_linear, r2_linear],
    ['⭐ Polynomial Regression (Degree 2)', mse_poly, r2_poly],
    ['🤖 Artificial Neural Network', mse_ann, r2_ann]
]

print("\nComparison of Models : \n")
print(tabulate(results, headers=['Model', 'MSE', 'R² Score'], tablefmt='fancy_grid'))
```
